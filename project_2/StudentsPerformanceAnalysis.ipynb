### 1.9 · Missing Values Detection
Count null values per column and visualise them with a heatmap-style chart.

### 1.10 · Duplicate Rows Detection & Removal
Check for exact duplicate rows and remove them if present.

```python
dup_count = df.duplicated().sum()
print(f'Duplicate rows found: {dup_count}')

if dup_count > 0:
    print('\n--- Sample duplicates ---')
    display(df[df.duplicated(keep=False)].head(10))
    df.drop_duplicates(inplace=True)
    df.reset_index(drop=True, inplace=True)
    print(f'\nDuplicates removed. New shape: {df.shape}')
else:
    print('No duplicates detected.')
```

### 1.11 · Unwanted Whitespace Detection & Fix
Check every object column for leading/trailing whitespace. If found, strip it.

```python
obj_cols = df.select_dtypes(include='object').columns
whitespace_issues = {}

for col in obj_cols:
    raw = df[col].unique()
    stripped = df[col].str.strip().unique()
    diff = len(raw) - len(stripped)
    if diff > 0:
        whitespace_issues[col] = diff

if whitespace_issues:
    print('Whitespace issues found:')
    for col, cnt in whitespace_issues.items():
        print(f'   {col}: {cnt} extra unique values due to whitespace')
    # Fix
    for col in obj_cols:
        df[col] = df[col].str.strip()
    print('\nAll leading/trailing whitespace removed.')
else:
    print('No whitespace issues detected.')
```

### 1.12 · Spelling & Casing Consistency Check
Print all unique values for each categorical column. Look for misspellings, inconsistent casing, or unexpected categories.

```python
for col in obj_cols:
    print(f'--- {col} ({df[col].nunique()} unique) ---')
    print(sorted(df[col].unique()))
    print()
```

**Observation:** All categorical values are consistently lowercase with no misspellings. No corrections needed.

### 1.13 · Format Standardisation
Rename columns to snake_case for cleaner code. Ensure score columns are integer type.

```python
# Rename columns to snake_case
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_').str.replace('/', '_')
print('Renamed columns:')
print(list(df.columns))

# Ensure score columns are integer
score_cols = ['math_score', 'reading_score', 'writing_score']
for col in score_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce').astype('Int64')

print('\nColumns renamed & score dtypes verified.')
df.dtypes
```

### 1.14 · Feature Engineering
Create a **total_score** column (sum of all three exams) and an **average_score** column to enable cross-student comparison.

```python
df['total_score'] = df['math_score'] + df['reading_score'] + df['writing_score']
df['average_score'] = (df['total_score'] / 3).round(2)

# Assign a pass/fail label (pass = average >= 50)
df['result'] = df['average_score'].apply(lambda x: 'Pass' if x >= 50 else 'Fail')

# Assign letter grades
def assign_grade(avg):
    if avg >= 90: return 'A'
    elif avg >= 80: return 'B'
    elif avg >= 70: return 'C'
    elif avg >= 60: return 'D'
    elif avg >= 50: return 'E'
    else: return 'F'

df['grade'] = df['average_score'].apply(assign_grade)

print('New columns: total_score, average_score, result, grade')
df.head()
```

### 1.15 · Outlier Detection (IQR Method)
Use the Inter-Quartile Range fence to identify outliers in each score column. Visualise with box plots.

```python
def detect_outliers_iqr(series):
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = series[(series < lower) | (series > upper)]
    return lower, upper, outliers

for col in score_cols:
    lower, upper, outliers = detect_outliers_iqr(df[col])
    print(f'{col}:  IQR fence [{lower:.1f}, {upper:.1f}]  |  Outliers: {len(outliers)}')
```

**Visualise outliers** using box plots to understand their position.

```python
fig = px.box(
    df, y=score_cols,
    title='Box Plots — Score Columns (Outlier Detection)',
    labels={'value': 'Score', 'variable': 'Subject'},
    color_discrete_sequence=px.colors.qualitative.Set2
)
fig.update_layout(template='plotly_dark', title_x=0.5)
fig.show()
```

### 1.16 · Outlier Treatment (Capping / Winsorization)
Cap any values outside the IQR fences to the fence boundaries. This preserves the row while limiting extreme influence.

```python
for col in score_cols:
    lower, upper, _ = detect_outliers_iqr(df[col])
    before = len(df[(df[col] < lower) | (df[col] > upper)])
    df[col] = df[col].clip(lower=lower, upper=upper)
    after = len(df[(df[col] < lower) | (df[col] > upper)])
    print(f'{col}: capped {before} outliers  →  remaining: {after}')

# Recalculate derived columns after capping
df['total_score'] = df['math_score'] + df['reading_score'] + df['writing_score']
df['average_score'] = (df['total_score'] / 3).round(2)
df['result'] = df['average_score'].apply(lambda x: 'Pass' if x >= 50 else 'Fail')
df['grade'] = df['average_score'].apply(assign_grade)

print('\nOutliers capped & derived columns recalculated.')
```

### 1.17 · Post-Cleaning Verification
Final sanity check after all cleaning steps.

```python
print('=== Post-Cleaning Summary ===')
print(f'Shape           : {df.shape}')
print(f'Nulls remaining : {df.isnull().sum().sum()}')
print(f'Duplicates      : {df.duplicated().sum()}')
print(f'\nDtypes:\n{df.dtypes}')
print('\nDataset is clean and ready for analysis.')
```

### Q9 — Is there a combined effect of lunch type + test prep on performance?
Examine the interaction between these two key factors.

- **Free/reduced lunch + No prep course** is the lowest-performing combination.  
- Even within the free/reduced lunch group, completing the prep course significantly boosts scores, suggesting the course can partially offset socio-economic disadvantage.
